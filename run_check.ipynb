```python
from query_helper import AudioRAGQuerier

rag = AudioRAGQuerier(rag_store="./rag_store")

query_text = "growing up in New York"
results = rag.semantic_search(query_text=query_text, n_results=3)

print(f"=== 1. SEMANTIC SEARCH RESULTS ('{query_text}') ===")
print(f"Total Hits: {len(results)}\n")

for i, res in enumerate(results, start=1):
    print(f"Result #{i}:")
    print(f"  Segment ID   : {res['segment_id']}")
    print(f"  Timestamp    : [{res['start_sec']:.1f}s - {res['end_sec']:.1f}s]")
    print(f"  Speaker      : {res['speaker_label']}")
    print(f"  Distance     : {res['distance']}")
    print(f"  Audio Blob   : {res['audio_blob']}")
    print(f"  Transcript   : \"{res['transcript']}\"")
    print("-" * 50)
```

```python

rag = AudioRAGQuerier(rag_store="./rag_store")

ordered_segments = rag.get_ordered_segments()

print("=== 2. CHRONOLOGICAL TRANSCRIPT ===")
print(f"Total Segments Found: {len(ordered_segments)}\n")

for seg in ordered_segments:
    start_fmt = f"{seg['start_sec']:.1f}s"
    end_fmt = f"{seg['end_sec']:.1f}s"
    speaker = f"Speaker {seg['speaker']}" if seg['speaker'] is not None else "Unknown"
    
    print(f"[{start_fmt.rjust(6)} -> {end_fmt.rjust(6)}] {speaker}: {seg['text']}")
```
