```python
def merge_speaker_segments(segments):
    if not segments:
        return []

    merged = [dict(segments[0])]

    for current in segments[1:]:
        previous = merged[-1]

        # Merge purely based on speaker label matching
        if previous['label'] == current['label']:
            previous['end'] = max(previous['end'], current['end'])
            previous['end_sample'] = max(previous['end_sample'], current['end_sample'])
        else:
            merged.append(dict(current))

    return merged

result = merge_speaker_segments(segments)

```

```python
from pydub import AudioSegment

def cut_wav_audio(input_file, output_file, start_sec, end_sec):
    # milliseconds
    start_ms = start_sec * 1000
    end_ms = end_sec * 1000
    
    # Load, slice, and save
    audio = AudioSegment.from_wav(input_file)
    clipped_audio = audio[start_ms:end_ms]
    clipped_audio.export(output_file, format="wav")

```

```python
from pathlib import Path

path = Path("./data/segments")

wavs = ["./data/segments/" + f.name for f in path.iterdir()]
print(wavs)
```

```python
import os
from dotenv import load_dotenv
load_dotenv()

from sarvamai import SarvamAI

client = SarvamAI(
    api_subscription_key=os.getenv("SARVAM_API_KEY"),
)

def transcribe(input_file: str):
    with open(input_file, "rb") as f:
        response = client.speech_to_text.transcribe(
            file=f,
            model="saaras:v4",
            language_code="en-IN",
            mode="transcribe",
            # sample_rate=16000,
        )

    return response

```

```python
out = f"{wavs[0]}.txt"
trans = transcribe(wavs[0])

with open(out, "w") as f:
    f.write(trans.transcript)
```

```python
for file in wavs:
    try:
        output_file = f"{file}.txt"
        trans = transcribe(file)
        
        with open(output_file, "w") as f:
            f.write(trans.transcript)
    except Exception as e:
        print(f"error {e}")
```

```python
rag_store = "./rag_store"

import chromadb
from chromadb.utils import embedding_functions

# 1. Initialize persistent Chroma client (saves data to a local folder)
# This directory will be created in your current working path
chroma_client = chromadb.PersistentClient(path=rag_store)

# 2. Set up the SentenceTransformer embedding function for Chroma
# Chroma handles the encoding internally when you use this utility wrapper
sentence_transformer_ef = embedding_functions.SentenceTransformerEmbeddingFunction(
    model_name="google/embeddinggemma-300m"
)
```

```python
collections_to_delete = ["audios", "audio_segments", "transcribe_segments", "audio_attributes"]

for collection_name in collections_to_delete:
    try:
        chroma_client.delete_collection(name=collection_name)
        print(f"Successfully deleted: {collection_name}")
    except ValueError:
        # Avoids crashing if the collection doesn't exist
        print(f"Collection '{collection_name}' not found. Skipping...")
```

```python
import base64
import chromadb

# Create or get your target collection
collection = chroma_client.get_collection("trial_segments")

# 2. Function to read a file and convert it to a Base64 string
def file_to_blob_string(file_path: str) -> str:
    with open(file_path, "rb") as file:
        binary_data = file.read()
        # Encode to base64 bytes and decode to a utf-8 string
        base64_string = base64.b64encode(binary_data).decode("utf-8")
    return base64_string

# 3. Path to the file you want to store
file_to_store = wavs[0] # Can be an image, PDF, zip, etc.

# Convert file to BLOB string representation
blob_content = file_to_blob_string(file_to_store)

# 4. Insert the BLOB into ChromaDB
collection.add(
    ids=["file_id_001"],
    documents=[blob_content], # Stored as the primary text document
    metadatas=[{
        "filename": file_to_store,
        "type": "wav",
        "description": ""
    }]
)
print("File successfully stored as a blob in ChromaDB.")
```
